#Practical 03: Apply Filtering, Sorting, Aggregation, and Group By Operations on Real-World Dataset.

```python
# Load necessary libraries for data manipulation
import pandas as pd
import numpy as np
```

**Step 1: Create a Dataframe/Records**

```python
# Define the dictionary containing employee attributes and records
data = {
    "Emp_ID" : [101, 102, 103, 104, 105, 106, 107, 108],
    "Name" : ["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department" : ["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City" : ["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000, 38000, 52000, 48000, 40000, 51000, 60000, 42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

```python
# Initialize a DataFrame object using the dictionary data
df = pd.DataFrame(data);
```

```python
# Present the generated DataFrame contents
df
```

**Step 2: Filter Data**

2.1 Display employee from IT Department

```python
# Retrieve and show rows where the department matches 'IT'
df[df["Department"] == "IT"]
```

2.2 Display Employees with salary greater than 20000

```python
# Query the DataFrame for records with compensation exceeding 20000
df[df['Salary'] > 20000]
```

2.3 Display Employee from Surat

```python
# Filter the data to list only those based in Surat
df[df['City'] == "Surat"]
```

2.4 Apply Multiple Condition

```python
# Extract IT department staff earning more than 50000 using compound criteria
df[(df['Department'] == "IT") & (df['Salary'] > 50000)]
```

**Step 3: Sorting Data**

3.1 Sort Salary in Ascending Order

```python
# Arrange the records in ascending order of salary
df.sort_values(by="Salary")
```

3.2 Sort Salary in descending order

```python
# Order the rows from highest to lowest compensation
df.sort_values(by="Salary", ascending=False)
```

3.3 Sort Department Wise Salary

```python
# Organize primarily by department name, then sort salaries downward within each group
df.sort_values(by=["Department", "Salary"], ascending=[True, False])
```

**Step 4: Perform Aggregation**

4.1 Calculate Total Salary

```python
# Compute the cumulative salary of all employees combined
df["Salary"].sum()
```

4.2 Calculate the avg salary

```python
# Determine the mean compensation value of the workforce
df["Salary"].mean()
```

4.3 Find Highest Salary

```python
# Identify the maximum earnings value in the dataset
df['Salary'].max()
```

4.4 Find Lowest Salary

```python
# Find the minimum earnings value recorded in the dataset
df["Salary"].min()
```

4.5 Count Employees

```python
# Obtain the total number of employee records based on ID count
df['Emp_ID'].count()
```

**Step 5:- Perform groupBy Operation**

5.1 Group Employee Based on department and Claculate the avg salary of each department

```python
# Segment by department and compute the mean salary for each segment
df.groupby('Department')['Salary'].mean()
```

5.2 Calculate the total salary paid by department

```python
# Segment by department to aggregate the sum of salaries distributed
df.groupby('Department')['Salary'].sum()
```

5.3 Display the highest salary in each department

```python
# Group the rows by department and extract the top salary in each group
df.groupby('Department')['Salary'].max()
```

5.4 Count Employees Department Wise

```python
# Group the records by department to tally the headcount of each
df.groupby('Department')['Salary'].count()
```

5.5 Calculate the average salary of employees in each city

```python
# Segment the dataset by city to calculate the average employee salary
df.groupby('City')['Salary'].mean()
```

**Step 6 : Multiple Aggregation Function**

```python
# Produce multiple summary metrics for salary across each distinct department
df.groupby('Department')['Salary'].agg(['count', 'sum', 'mean', 'min', 'max'])
```

**Step 7 : save the result**

```python
# Export the structured DataFrame to a local CSV file, omitting row indices
df.to_csv("Employee_Data.csv", index=False)
print("Dataset Saved Successfully")
```
