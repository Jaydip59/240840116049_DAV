# **Practical 01: Study the Data Analytics lifecycle and perform data acquisition from CSV,Excel, and JSON files using Pandas.**

**Step1: Importing Library.**

```python
# Import Pandas Library.
import pandas as pd
```

**Step:2 Create a DataFrame.**

```python
data = {
    "ID" : [101,102,103,104,105,106,107],
    "Name" : ["Rahul", "Priya", "Amit", "Riya", "Karan", "Krish", "Harry"],
    "Age" : [20,21,19,20,22,24,28],
    "City" : ["Surat", "Navsari", " Valsad", "Bharuch", "Surat" , "Gandhinagar", "Ahemdabad"],
    "Marks":  [85,92,78,88,81,84,90]
}
```

**Step:3 Storing a DataFrame.**

```python
# Store a DataFrame in df variable.
df = pd.DataFrame(data)
```

**Step:4 Display a DataFrame.**

```python
# Display the pd.DataFrame
df
```

**Step:5 Data Converting in CSV.**

```python
# Save Data as a CSV File
df.to_csv("student.csv", index=False) # index=False for index value remove
print("CSV file created successfully.") # dor the Display msg CSV file created...
```

**Step:6 Data Converting in EXCEL.**

```python
# Save Data as a EXCEL File
df.to_excel("student.xlsx", index=False) # index=False for index value remove
print("EXCEL file created successfully.") # dor the Display msg EXCEL file created...
```

**Step:7 Data Converting in JSON.**

```python
# Save Data as a JSON File
df.to_json("student.json",orient="records", index=False) #  orinets:- create a list of JSON Objects index=False for index value remove
print("JSON file created successfully.") # dor the Display msg JSON file created...
```
